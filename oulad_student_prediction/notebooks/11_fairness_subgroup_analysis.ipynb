# Phase 14: Fairness & Subgroup Predictive Performance Analysis

**Study**: *Student Performance Prediction Using Machine Learning and Explainable AI: An Experimental Study Using the Open University Learning Analytics Dataset (OULAD)*

### Phase 14 Methodological Constraints & Guarantees:
1. **Frozen Model Architecture**: Explains the exact, unmodified Scenario $S_2$ `HistGradientBoostingClassifier` trained in Phase 11 and evaluated in Phase 12. Zero retraining, zero hyperparameter adjustments, zero feature changes.
2. **Frozen Decision Threshold**: Evaluates performance using strictly the locked operational threshold $\theta^* = 0.45$. No subgroup-specific threshold optimization or adjustment is permitted.
3. **Locked Test Partition**: Evaluated out-of-sample strictly on the holdout test set ($N = 4,889$ observations across $4,318$ unique students).
4. **Small-Cell Rule**: Minimum sample size of $n \ge 100$ test observations is enforced for primary metric estimation. Subgroups with $n < 100$ are reported as having insufficient sample size.
5. **Student-Level Cluster Bootstrap**: 95% confidence intervals are generated via $B = 1,000$ block cluster resamples of `id_student` (`RANDOM_SEED = 42`) to account for repeat course enrollments.
6. **Objective & Neutral Reporting**: Subgroup differences are characterized descriptively alongside target prevalence (base rates) without moralistic fairness labels, subgroup rankings, or causal assertions.
