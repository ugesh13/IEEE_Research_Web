# Phase 15: Statistical Robustness and Sensitivity Analysis

**Study**: *Student Performance Prediction Using Machine Learning and Explainable AI: An Experimental Study Using the Open University Learning Analytics Dataset (OULAD)*

### Phase 15 Methodological Guarantees and Constraints:
1. **Frozen Primary Experiment**: The final model specification ($S_2$ `HistGradientBoostingClassifier`, $\theta^* = 0.45$, unweighted) and its holdout test metrics ($N = 4,889$, Accuracy = $75.95\%$, Recall = $82.38\%$, Precision = $74.68\%$, $F_1 = 0.7834$, ROC-AUC = $0.8406$, PR-AUC = $0.8459$, Brier = $0.1638$; $\text{TP} = 2,127, \text{FN} = 455, \text{TN} = 1,586, \text{FP} = 721$) remain completely locked and are **NOT** replaced or retuned.
2. **Strict Test Set Protection**: The holdout test set is never used for hyperparameter tuning, threshold selection, model selection, or feature selection.
3. **Zero Clustering Leakage**: Rigorous student-level grouping is enforced across all 5 random partition seeds (seeds 7, 21, 42, 84, 123), with zero student overlap verified across train, validation, and test splits.
4. **Comprehensive Sensitivity Scope**:
   - **Analysis A**: Decision threshold sensitivity across grid $\theta \in [0.20, 0.80]$.
   - **Analysis B**: Partition and random seed robustness across 5 seeds.
   - **Analysis C**: Bootstrap resampling sensitivity (student-level cluster bootstrap vs observation bootstrap).
   - **Analysis D**: Class-weight sensitivity (`class_weight=None` vs `class_weight='balanced'`).
   - **Analysis E**: Scenario progression robustness ($S_0 \rightarrow S_1 \rightarrow S_2$) across partitions.
