# Phase 6: Feature Engineering & Temporal Scenario Construction (S0, S1, S2)

**Study**: *Student Performance Prediction Using Machine Learning and Explainable AI: An Experimental Study Using the Open University Learning Analytics Dataset (OULAD)*

### Phase 6 Methodological Standards:
1. **Strict Independent Calculation**: Features are derived directly from raw tables using scenario-specific temporal cutoffs (no copying full-semester aggregates).
2. **Nested Scenario Architecture**:
   - **S0 (Baseline Pre-Course, Day $\le 0$)**: Demographics, course attributes, registration timing.
   - **S1 (Early Academic, Day $\le 30$)**: $S_0$ + assignments submitted on or before Day 30.
   - **S2 (Early Learning Behavior, Day $\le 30$)**: $S_1$ + VLE interactions logged on or before Day 30.
3. **Zero Target Leakage**: Final exams excluded; `date_unregistration` excluded; post-cutoff data barred.
4. **Score Representation**: Unsubmitted assessment scores remain `NaN` in raw matrices with explicit binary submission flag (`has_early_assessment_submission_d30`), deferring imputation to model training pipelines (Phase 10).

```python
import os
import sys
import pandas as pd
import numpy as np

PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), ".."))
SRC_DIR = os.path.join(PROJECT_ROOT, "src")
RAW_DIR = os.path.join(PROJECT_ROOT, "data", "raw")
PROCESSED_DIR = os.path.join(PROJECT_ROOT, "data", "processed")
TABLES_DIR = os.path.join(PROJECT_ROOT, "results", "tables")

if SRC_DIR not in sys.path:
    sys.path.append(SRC_DIR)

from feature_engineering import build_temporal_scenarios

s0, s1, s2 = build_temporal_scenarios(RAW_DIR, PROCESSED_DIR, TABLES_DIR)

print(f"S0 Shape: {s0.shape}")
print(f"S1 Shape: {s1.shape}")
print(f"S2 Shape: {s2.shape}")
```
