# Phase 13: Explainable AI / SHAP Analysis for Student Performance Prediction

**Study**: *Student Performance Prediction Using Machine Learning and Explainable AI: An Experimental Study Using the Open University Learning Analytics Dataset (OULAD)*

### Phase 13 Methodological Constraints & Guarantees:
1. **Interpretation Stage Only**: SHAP is strictly utilized to explain the predictions of the frozen final binary classification model. Zero features are added or removed, no hyperparameters or thresholds are tuned, and no retraining occurs based on SHAP.
2. **Frozen Final Model Verified**: Explains Scenario $S_2$ `HistGradientBoostingClassifier(learning_rate=0.05, max_iter=200, max_leaf_nodes=31, min_samples_leaf=20, class_weight=None, random_state=42)` operating at $\theta^* = 0.45$.
3. **Training Background Isolation**: The reference background ($N_{\text{bg}} = 100$) was drawn strictly from the training partition (`RANDOM_SEED = 42`). Zero validation or test data were used as reference background.
4. **Target Directionality**: Positive SHAP values indicate an increased predicted log-odds/probability of being **At-Risk** (`is_at_risk = 1`), while negative SHAP values indicate an increased probability of being **Successful** (`is_at_risk = 0`).
5. **Dual Importance Hierarchy**: Global feature importance is reported both at the transformed feature level (68 columns) and aggregated by original raw predictor (29 features) via $\sum_{k \in \text{Transformed}(X_j)} \mathbb{E}[|\phi_k|]$.
6. **Non-Causal Interpretation**: Attributions reflect statistical associations learned by the tree ensemble within OULAD; they do not infer causality or interventional impact.
