```python
import os
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), ".."))
SRC_DIR = os.path.join(PROJECT_ROOT, "src")
PROCESSED_DIR = os.path.join(PROJECT_ROOT, "data", "processed")
FIGURES_DIR = os.path.join(PROJECT_ROOT, "results", "figures")

if SRC_DIR not in sys.path:
    sys.path.append(SRC_DIR)

from eda_figures import generate_all_eda_figures

# Execute publication figure pipeline
generate_all_eda_figures(PROCESSED_DIR, FIGURES_DIR)
```

```python
# Load and verify S0, S1, S2 descriptive statistics
s0 = pd.read_csv(os.path.join(PROCESSED_DIR, 'scenario_s0.csv'))
s1 = pd.read_csv(os.path.join(PROCESSED_DIR, 'scenario_s1.csv'))
s2 = pd.read_csv(os.path.join(PROCESSED_DIR, 'scenario_s2.csv'))

print("=== Demographic Risk Distribution by Highest Education ===")
print(s0.groupby('highest_education')['is_at_risk'].agg(['count', 'mean']).rename(columns={'mean': 'at_risk_rate'}).mul([1, 100]).round(2))

print("\n=== Early Assessment Scores Summary (Submitting Students) ===")
print(s1.groupby('is_at_risk')['early_avg_score_d30'].describe().round(2))

print("\n=== Early VLE Clicks Summary ===")
print(s2.groupby('is_at_risk')['early_vle_total_clicks_d30'].describe().round(1))
```
